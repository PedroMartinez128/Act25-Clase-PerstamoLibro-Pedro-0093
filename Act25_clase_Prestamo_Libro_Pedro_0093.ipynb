```python
class PrestamoLibro:
    def __init__(self):
        # Atributos inicializados vacíos
        self.usuario = None
        self.libro = None
        self.fecha = None

    # Función 1: Leer_Datos (Pide información al usuario por teclado)
    def leer_datos(self, nombre_instancia):
        print(f"--- Captura de datos para {nombre_instancia} ---")
        self.usuario = input("Escribe el nombre del Usuario: ")
        self.libro = input("Escribe el título del Libro: ")
        self.fecha = input("Escribe la Fecha del préstamo: ")
        print(f"¡Datos de {nombre_instancia} guardados!\n")

    # Función 2: Mostrar_datos (Acción de imprimir los resultados)
    def mostrar_datos(self):
        print("-------------------------------")
        print(f"DATOS REGISTRADOS:")
        print(f"Usuario: {self.usuario}")
        print(f"Libro:   {self.libro}")
        print(f"Fecha:   {self.fecha}")
        print("-------------------------------\n")

# --- Creación de los objetos (Instancias) ---
biblio1 = PrestamoLibro()
biblio2 = PrestamoLibro()

# --- Llamada a las funciones ---

# Primero capturamos los datos de ambos
biblio1.leer_datos("Biblio1")
biblio2.leer_datos("Biblio2")

# Luego mostramos la información procesada
print("PROCESANDO INFORMACIÓN FINAL...\n")
biblio1.mostrar_datos()
biblio2.mostrar_datos()
print("Programa realizado por Pedro Martinez NL = 34")
```
